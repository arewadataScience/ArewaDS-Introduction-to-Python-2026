## Lists

A **list** is an ordered sequence of data values (called **items** or **elements**).

Some real-world examples:
- Shopping list for the grocery store
- To-do list
- Roster for an athletic team
- Guest list for an event
- A recipe (list of instructions)
- A text document (list of lines)

Each item in a list has a unique **index** starting at `0` up to `len(list) - 1`.

### Replacing Elements in a List

```python

example = [1, 2, 3, 4]
example[3] = 0  # replace item at index 3
example
```

```python

# Replace each number with its square
numbers = [2, 3, 4, 5]
for i in range(len(numbers)):
    numbers[i] = numbers[i] ** 2
numbers
```

```python

# Using split to extract words then uppercasing them
sentence = "This example has five words."
words = sentence.split()
for i in range(len(words)):
    words[i] = words[i].upper()
words
```

### List Methods for Inserting and Removing Elements

```python

# insert(index, value)
example = [1, 2]
example.insert(1, 10)
example.insert(3, 25)
example
```

```python

# append(value) and extend(iterable)
example = [1, 2]
example.append(3)
example.extend([11, 12, 13])
example
```

```python

# pop() removes and returns an element
example = [1, 2, 10, 11, 12, 13]
last = example.pop()      # removes last
first = example.pop(0)    # removes first
example, last, first
```

### Sorting a List

```python

example = [4, 2, 10, 8]
example.sort()   # in-place
example
```

#### Mutator Methods and `None`

```python

a_list = [3, 1, 2]
result = a_list.sort()   # returns None because it's a mutator
print("Returned:", result)
print("Sorted list:", a_list)
```

### Aliasing and Side Effects

```python

first = [10, 20, 30]
second = first       # aliasing: both refer to the same list object
first[1] = 99
first, second, (first is second)
```

```python

# Prevent aliasing by copying
first = [10, 99, 30]
third = list(first)  # or first[:]
first[1] = 100
first, third, (first is third)
```

### Equality: Object Identity vs Structural Equivalence

```python

first = [20, 30, 40]
second = first
third = list(first)  # or first[:]

(first == second, first == third, first is second, first is third)
```

### Example: Find the Median of Numbers in a File

**Task:** Read numbers from a text file and print the median.

> For live class, replace `"numbers.txt"` with a path to a small file in the same folder.

```python

# Demo-friendly version: we'll simulate file content to avoid IO errors in class
text = "3 10 6 2 8\n5 9\n"
numbers = []
for line in text.splitlines():
    for word in line.split():
        numbers.append(float(word))

numbers.sort()
mid = len(numbers) // 2
if len(numbers) % 2 == 1:
    median = numbers[mid]
else:
    median = (numbers[mid] + numbers[mid - 1]) / 2
numbers, median
```

## Tuples

A **tuple** resembles a list but is **immutable** (cannot be modified). Use parentheses `()`.

```python

fruits = ("apple", "banana")
meats = ("fish", "poultry")
food = meats + fruits
veggies = ["celery", "beans"]
tuple_veggies = tuple(veggies)
fruits, meats, food, tuple_veggies
```

```python

def odd(x):
    """Returns True if x is odd, else False."""
    if x % 2 == 1:
        return True
    else:
        return False

odd(5), odd(6)
```

## Dictionaries

A **dictionary** organizes information by **association** using **key → value** pairs.
- Also called *tables* or *associative arrays*.
- Keys are immutable (e.g., strings, integers, tuples).
- Values can be any type.

### Dictionary Literals

```python

phone_book = {"Savannah": "476-3321", "Nathaniel": "351-7743"}
person = {"Name": "Molly", "Age": 18}
empty = {}
phone_book, person, empty
```

### Adding Keys and Replacing Values

```python

info = {}
info["name"] = "Sandy"
info["occupation"] = "hacker"
info["occupation"] = "manager"  # replace
info
```

### Accessing Values

```python

info = {"name": "Sandy"}
name = info["name"]
missing = info.get("job", None)  # safe access with default
name, missing
```

### Removing Keys

```python

info = {"name": "Sandy", "occupation": "manager"}
removed1 = info.pop("job", None)       # default when absent
removed2 = info.pop("occupation")      # removes the key
info, removed1, removed2
```

### Traversing a Dictionary

```python

grades = {90: 'A', 80: 'B', 70: 'C'}

# Method 1: iterate keys
pairs1 = [(k, grades[k]) for k in grades]

# Method 2: items()
pairs2 = list(grades.items())

# Sorted by key
sorted_keys = list(grades.keys())
sorted_keys.sort()
pairs_sorted = [(k, grades[k]) for k in sorted_keys]

pairs1, pairs2, pairs_sorted
```

### Example: Hexadecimal System Revisited

```python

hex_to_bin = {
    '0':'0000', '1':'0001', '2':'0010', '3':'0011',
    '4':'0100', '5':'0101', '6':'0110', '7':'0111',
    '8':'1000', '9':'1001', 'A':'1010', 'B':'1011',
    'C':'1100', 'D':'1101', 'E':'1110', 'F':'1111'
}

def convert(number, table):
    """Builds and returns the base-2 representation of a hexadecimal string."""
    binary = ""
    for digit in number:
        binary = table[digit] + binary
    return binary

convert("35A", hex_to_bin)
```

### Example: Finding the Mode of a List of Words

```python

text = "red red blue green green green BLUE"
# Normalize to uppercase and split
words = [w.upper() for w in text.split()]

# Frequency dictionary
freq = {}
for w in words:
    count = freq.get(w, 0)
    freq[w] = count + 1

# Find mode
the_max = max(freq.values())
mode = [k for k, v in freq.items() if v == the_max][0]
freq, mode
```
